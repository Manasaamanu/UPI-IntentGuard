```python
import pandas as pd

df = pd.read_csv("data/raw/synthetic_transactions.csv")

user_id = "U0000"

user_history = df[
    df["user_id"].astype(str) == user_id
].copy()

user_history["timestamp"] = pd.to_datetime(
    user_history["timestamp"]
)

user_history = user_history.sort_values("timestamp")

print("USER:", user_id)

print("\nAVERAGE PAYMENT:")
print(round(user_history["amount"].mean(), 2))

print("\nFAMILIAR RECIPIENTS:")
print(user_history["recipient_id"].value_counts().head(10))

print("\nMOST USED CATEGORY:")
print(user_history["category"].value_counts())

print("\nLATEST TRANSACTION:")
display(user_history.tail(1)[[
    "amount",
    "recipient_id",
    "timestamp",
    "category",
    "device_type",
    "location"
]])
```

```python
import pandas as pd

df = pd.read_csv("data/raw/synthetic_transactions.csv")

print("Shape:", df.shape)
print("\nFraud distribution:")
print(df["is_fraud"].value_counts())

print("\nFraud percentage:")
print(df["is_fraud"].value_counts(normalize=True) * 100)

print("\nFraud rate by scenario:")
print(df.groupby("scenario")["is_fraud"].agg(["count", "mean"]))

print("\nAverage amount by class:")
print(df.groupby("is_fraud")["amount"].mean())

print("\nDevice distribution by class:")
print(pd.crosstab(df["device_type"], df["is_fraud"], normalize="columns"))

print("\nLocation distribution by class:")
print(pd.crosstab(df["location"], df["is_fraud"], normalize="columns"))
```

```python
import pandas as pd
import numpy as np

df = pd.read_csv("data/raw/synthetic_transactions.csv")

print("1. Missing values:")
print(df.isnull().sum())

print("\n2. Duplicate transactions:")
print(df["transaction_id"].duplicated().sum())

print("\n3. Duplicate complete rows:")
print(df.duplicated().sum())

print("\n4. Fraud distribution across users:")
user_fraud = df.groupby("user_id")["is_fraud"].agg(
    total_transactions="count",
    fraud_transactions="sum"
)
print(user_fraud.describe())

print("\n5. Fraud rate by category:")
print(
    df.groupby("category")["is_fraud"]
      .agg(["count", "mean"])
      .sort_values("mean", ascending=False)
)

print("\n6. Fraud rate by device:")
print(df.groupby("device_type")["is_fraud"].mean())

print("\n7. Fraud rate by location:")
print(df.groupby("location")["is_fraud"].mean())

print("\n8. Columns available:")
print(df.columns.tolist())

print("\n9. Check scenario leakage:")
print("Scenario column exists:", "scenario" in df.columns)
```
